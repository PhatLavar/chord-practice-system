# P1 · Compact Transformer teacher F
Project-specific Transformer with the existing chord components. Run the overfit check before the full balanced-sampling trial. Enable Internet and GPU.

```python
from pathlib import Path
import subprocess, sys, os, json

project = Path("/kaggle/working/chord-practice-system")
if not project.exists():
    subprocess.run(
        ["git", "clone", "https://github.com/PhatLavar/chord-practice-system.git", str(project)],
        check=True,
    )
os.chdir(project)
subprocess.run(["git", "pull", "--ff-only", "origin", "main"], check=True)
subprocess.run([sys.executable, "-m", "pip", "install", "--no-deps", "-e", "."], check=True)
print(subprocess.check_output(["git", "rev-parse", "HEAD"], text=True))
import torch, numpy

print("torch", torch.__version__, "numpy", numpy.__version__, "CUDA", torch.cuda.is_available())
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
# Official inputs. Downloaded archives are read without extraction.
from urllib.request import urlretrieve

raw = project / "data/raw"
raw.mkdir(parents=True, exist_ok=True)
sources = {
    "index.csv": "https://www.dropbox.com/s/o0olz0uwl9z9stb/billboard-2.0-index.csv?dl=1",
    "full.tar.xz": "https://www.dropbox.com/s/t390alzrkx0c9yt/billboard-2.0.1-lab.tar.xz?dl=1",
    "billboard-2.0-chordino.tar.xz": "https://www.dropbox.com/s/e9dm23vbawg9dsw/billboard-2.0-chordino.tar.xz?dl=1",
}
for name, url in sources.items():
    if not (raw / name).exists():
        temporary = raw / (name + ".part")
        urlretrieve(url, temporary)
        temporary.replace(raw / name)
cache = project / "data/processed/p1"
if not (cache / "metadata.json").exists():
    subprocess.run([sys.executable, "-m", "acr.data.billboard.prepare"], check=True)
subprocess.run([sys.executable, "-m", "unittest", "discover", "-s", "tests"], check=True)
```

```python
assert torch.cuda.is_available(), "Enable a Kaggle GPU."
experiment_root = project / "runs/p1-transformer"
overfit_run = experiment_root / "overfit"
subprocess.run(
    [
        sys.executable,
        "-m",
        "acr.training.frame",
        "--config",
        "configs/p1/p1-transformer-overfit.json",
        "--output",
        str(overfit_run),
        "--device",
        "cuda",
    ],
    check=True,
)
print((overfit_run / "metrics.json").read_text())
```

```python
RUN_FULL = False  # Set True after reviewing the overfit metrics.
transformer_run = experiment_root / "balanced"
if RUN_FULL:
    assert json.loads((overfit_run / "metrics.json").read_text())[
        "overfit_pass"
    ], "Review the failed pipeline check before full training."
    subprocess.run(
        [
            sys.executable,
            "-m",
            "acr.training.frame",
            "--config",
            "configs/p1/p1-transformer.json",
            "--output",
            str(transformer_run),
            "--device",
            "cuda",
        ],
        check=True,
    )
    subprocess.run(
        [
            sys.executable,
            "-m",
            "acr.evaluation.frame",
            "--run",
            str(transformer_run),
            "--output",
            str(transformer_run / "diagnostics"),
        ],
        check=True,
    )
    print((transformer_run / "metrics.json").read_text())
    print((transformer_run / "diagnostics/summary.json").read_text())
```
