# GuitarSet F/E: CQT transposition ablation
Training-only CQT shifts -5..+6 semitones with root-label transposition. Same player split, 30 epochs and update budgets as 02i. Clean validation and clean training metrics. Read docs/p2/guitarset/GUITARSET_TRANSPOSITION_EXPERIMENT.md. Enable Internet and GPU.


```python
from pathlib import Path
import subprocess, sys, os, json, hashlib

project = Path("/kaggle/working/chord-practice-system")
if not project.exists():
    subprocess.run(
        ["git", "clone", "https://github.com/PhatLavar/chord-practice-system.git", str(project)],
        check=True,
    )
os.chdir(project)
subprocess.run(["git", "pull", "--ff-only", "origin", "main"], check=True)
subprocess.run([sys.executable, "-m", "pip", "install", "--no-deps", "-e", "."], check=True)
print(subprocess.check_output(["git", "rev-parse", "HEAD"], text=True))
subprocess.run([sys.executable, "-m", "pip", "install", "-r", "requirements-audio.txt"], check=True)
import torch

print("torch", torch.__version__, "CUDA", torch.cuda.is_available())
subprocess.run([sys.executable, "-m", "unittest", "discover", "-s", "tests"], check=True)
```

```python
# Official pinned archives; about 696 MB total. No extraction to separate WAV files.

from acr.data.guitarset.audit import download
from acr.data.guitarset.cqt import load_cache

raw = Path("data/raw/guitarset")
download(raw)
config = "configs/p2/p2-guitarset-cqt-player.json"
cfg = json.loads(Path(config).read_text())
train_config = "configs/p2/p2-guitarset-cqt-transpose.json"
train_cfg = json.loads(Path(train_config).read_text())


def prepare_cache(path, smoke=False):
    if not path.exists():
        command = [
            sys.executable,
            "-m",
            "acr.data.guitarset.cqt",
            "--output",
            str(path),
            "--config",
            config,
        ]
        if smoke:
            command.append("--smoke")
        subprocess.run(command, check=True)
    # Checks archive identity recorded in cache, hashes, folds and train-only stats.
    songs, refs, meta = load_cache(path, cfg, smoke)
    print({"cache": str(path), "recordings": len(songs), "fold": cfg["fold"], "smoke": smoke})
    del songs, refs
```

```python
# Short CPU execution checks, separate from the full experiment.
smoke_cache = Path("data/processed/guitarset-cqt-player-smoke")
prepare_cache(smoke_cache, smoke=True)
revision = subprocess.check_output(["git", "rev-parse", "HEAD"], text=True).strip()
source_hashes = {
    p.as_posix(): hashlib.sha256(p.read_bytes()).hexdigest() for p in Path("src/acr").rglob("*.py")
}


def run_stage(cache, output, smoke=False):
    if output.exists():
        assert (
            output / "summary.json"
        ).exists(), (
            f"Incomplete run: {output}; preserve and inspect before choosing a fresh output."
        )
        for teacher in ("F", "E"):
            record = json.loads((output / teacher / "run.json").read_text())
            assert record["config"] == train_cfg and record["smoke"] == smoke
            assert record["git_revision"] == revision and record["source_sha256"] == source_hashes
        print("Reusing completed run", output)
        return
    command = [
        sys.executable,
        "-m",
        "acr.training.guitarset",
        "--cache",
        str(cache),
        "--output",
        str(output),
        "--config",
        train_config,
        "--device",
        "cpu" if smoke else "cuda",
    ]
    if smoke:
        command.append("--smoke")
    subprocess.run(command, check=True)


run_stage(smoke_cache, Path("runs/p2-guitarset-cqt-transpose-kaggle-smoke"), smoke=True)
```

```python
# Prepare 240 train + 60 validation recordings. No outer-test features.
cache = Path("data/processed/guitarset-cqt-player")
prepare_cache(cache)
```

```python
assert torch.cuda.is_available(), "Select a GPU for the full two-model pilot."
run_root = Path("runs/p2-guitarset-cqt-transpose")
run_stage(cache, run_root)
```

```python
summary = json.loads((run_root / "summary.json").read_text())
for teacher, result in summary.items():
    print(
        teacher,
        "best_epoch",
        result["best_epoch"],
        {
            name: round(values["score"], 4) if values["score"] is not None else None
            for name, values in result["metrics"].items()
        },
        "supported_oracle_event_accuracy",
        result["supported_oracle_event_accuracy"],
    )
print(
    "F: predicted frame timeline. E: gold event boundaries. Validation only; not published benchmark scores."
)

for teacher in ("F", "E"):
    control = json.loads(
        Path(f"reports/p2/guitarset-cqt/player-results/{teacher}_metrics.json").read_text()
    )
    print(
        teacher,
        "control tetrads",
        control["metrics"]["tetrads"]["score"],
        "augmented tetrads",
        summary[teacher]["metrics"]["tetrads"]["score"],
    )
```
